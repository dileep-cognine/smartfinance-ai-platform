```python
import re

def clean_sec_text(text: str) -> str:
    # Remove extra line breaks
    text = re.sub(r'\n+', ' ', text)

    # Remove multiple spaces
    text = re.sub(r'\s+', ' ', text)

    # Remove headings like Item 1A, page numbers
    text = re.sub(r'Item\s+\d+[A-Z]?', '', text, flags=re.IGNORECASE)

    # Remove unwanted symbols (optional)
    text = re.sub(r'[^a-zA-Z0-9.,;:%()\- ]', '', text)

    return text.strip()
```

```python
import os
import re

# Paths
input_dir = "data/raw/sec_filings/"
output_dir = "data/processed/cleaned/sec_filings/"

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Cleaning function
def clean_sec_text(text: str) -> str:
    text = re.sub(r'\n+', ' ', text)                 # remove line breaks
    text = re.sub(r'\s+', ' ', text)                 # remove extra spaces
    text = re.sub(r'Item\s+\d+[A-Z]?', '', text, flags=re.IGNORECASE)  # remove headings
    text = re.sub(r'[^a-zA-Z0-9.,;:%()\- ]', '', text)  # optional cleanup
    return text.strip()

# Process all files
for file_name in os.listdir(input_dir):
    if file_name.endswith(".txt"):
        input_path = os.path.join(input_dir, file_name)
        output_file = file_name.replace(".txt", "_cleaned.txt")
        output_path = os.path.join(output_dir, output_file)

        # Read
        with open(input_path, "r", encoding="utf-8") as f:
            raw_text = f.read()

        # Clean
        cleaned_text = clean_sec_text(raw_text)

        # Write
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(cleaned_text)

        print(f"Processed: {file_name} → {output_file}")

print("All SEC filings cleaned successfully!")
```
